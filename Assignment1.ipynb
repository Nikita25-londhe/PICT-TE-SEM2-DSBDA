```python
df['Discount']=df['Discount'].astype(str)
df['Discount']=df['Discount'].str.replace("% off","")
df['Discount']=df['Discount'].astype('float64')/100
df['Discount'].fillna(0,inplace=True)
```

```python
df['Price'].fillna(df['Price'].median(),inplace=True)
df['Avg Rating'].fillna(df['Avg Rating'].median(),inplace=True)
df['Price'].fillna(df['Price'].median())
df.dropna(subset=['Lens Detail','Memory Details'],inplace=True)
df['Battery Capacity'].fillna(df['Battery Capacity'].mode()[0],inplace=True)
df['Processor'].fillna(df['Processor'].mode()[0],inplace=True)
```

```python
df['Rating Counts']=df['Rating Count'].astype(str);
df['Rating Count']=df['Rating Count'].str.replace("Ratings","")
df['Rating Count']=df['Rating Count'].str.replace(",","")
df['Rating Count']=df['Rating Count'].astype('float64')
df['Rating Count'].fillna(0,inplace=True)
```

```python
df['Review Count']=df['Review Count'].astype(str);
df['Review Count']=df['Review Count'].str.replace("Reviews","")
df['Review Count']=df['Review Count'].str.replace(",","")
df['Review Count']=df['Review Count'].astype('float64')
df['Review Count'].fillna(0,inplace=True)
```

```python
for x in df.index:
    stream = str(df.loc[x, 'Warranty']).lower() 
    pos = stream.find('year')
    if pos == -1:
        df.loc[x, 'Warranty'] = 0
    else:
        year = stream[pos-3:pos].strip()
        df.loc[x, 'Warranty'] = int(year)
df['Warranty']=df['Warranty'].astype('float64')
```
